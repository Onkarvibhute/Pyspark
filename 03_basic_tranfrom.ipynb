```python
emp.printSchema()
```

```python
emp_casted=emp.select("emp_id","name","age",col("salary").cast("double"))
emp_casted.printSchema()
```

```python
#emp_taxed=emp_casted.withColumn("tax",emp_casted.salary*0.2)
#emp_taxed=emp_casted.withColumn("tax",col("salary")*0.2)
emp_taxed=emp_casted.withColumn("tax",expr("salary")*0.2)

emp_taxed.show()
```

```python
emp_lit=emp_taxed.withColumn("column_one",lit(1)).withColumn("column_two",lit("two"))
emp_lit.show()
```

```python
emp_1=emp_lit.withColumnRenamed("emp_id","e_id")
emp_1.show()
```

```python
emp_2=emp_lit.withColumnRenamed("column_two","column two")
emp_2.show()
```

```python
emp_dropped=emp_lit.drop("column_two")
emp_dropped.show()
```

```python
emp_filtered=emp_dropped.where("tax>10000")
emp_filtered.show()
```

```python
emp_limit=emp_filtered.limit(5)
emp_limit.show()
emp_limit.show(2)
```
